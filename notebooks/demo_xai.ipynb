# Faithful graph explanation demo
Demo này dùng scorer mẫu để kiểm tra pipeline. Thay `MockPathScorer` bằng `CallableScorerAdapter` khi chạy model thật.

```python
from mmxai.xai.demo_data import demo_graph, demo_schema
from mmxai.xai import EdgeContributionExplainer, MockPathScorer, XAIEvaluator
from mmxai.xai.visualize import format_summary, plot_explanation

graph, schema = demo_graph(), demo_schema()
query = 'person riding horse'
candidate_ids = ('1', '2')
scorer = MockPathScorer({'1': 0.72, '2': 0.69}, alpha=0.6, max_hops=2)
scores = scorer.score(query, candidate_ids, graph)
sorted(scores.items(), key=lambda row: -row[1].final_score)
```

```python
explanation = EdgeContributionExplainer(top_k=2).explain(query, '1', candidate_ids, graph, scorer)
metrics = XAIEvaluator(schema, random_repeats=30, seed=0).evaluate(
    query, '1', candidate_ids, graph, scorer, explanation
)
print(format_summary(explanation, metrics, graph))
metrics.to_dict()
```

```python
plot_explanation(explanation, graph)
```
